```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('../data/exemplo_vendas.csv', parse_dates=['data'])
df.head()
```

```python
# Visão geral dos dados
df.info()
df.describe(numeric_only=True)
```

## Faturamento por categoria

```python
por_categoria = df.groupby('categoria')['total'].sum().sort_values(ascending=False)
por_categoria
```

```python
fig = px.bar(por_categoria, title='Faturamento por categoria')
fig.show()
```

## Evolução mensal do faturamento

```python
df['mes'] = df['data'].dt.to_period('M').astype(str)
por_mes = df.groupby('mes')['total'].sum().reset_index()
px.line(por_mes, x='mes', y='total', markers=True, title='Faturamento por mês')
```
